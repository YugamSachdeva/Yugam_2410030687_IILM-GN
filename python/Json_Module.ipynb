# JSON Module in Python

## What is JSON?

**JSON (JavaScript Object Notation)** is a lightweight and human-readable data format used for **storing** and **exchanging data** between applications.

It is commonly used in:
- 🌐 Web APIs
- 💻 Web Applications
- ⚙️ Configuration Files
- 📦 Data Storage

JSON stores data in **key-value pairs**, making it easy for both humans and machines to read and understand.

### Example of JSON

```json
{
    "name": "Yugam",
    "age": 19,
    "course": "B.Tech CSE"
}
```

# JSON Module in Python

Python provides a built-in **`json`** module that allows you to:

- Convert Python objects into JSON format.
- Convert JSON data back into Python objects.
- Read JSON data from a file.
- Write JSON data to a file.

### Importing the JSON Module

```python
import json
```

The `json` module is useful because it helps Python applications:

- Exchange data with web servers (APIs).

## `json.dumps()` – Convert Python Object to JSON String

The **`json.dumps()`** function is used to convert a **Python object** into a **JSON-formatted string**.

````python
data={"name":"yugam","age":25,"city":"GN"}
a=json.dumps(data)
print(a)
print(type(a))
````

## `json.dump()` – Write JSON Data to a File

The **`json.dump()`** function is used to write **JSON data** to a file.


````python
with open("data.jason","w")as file:
    json.dump(data,file)
````

## Converting JSON to Python Objects (Deserialization)

**Deserialization** (also called **decoding** or **loading**) is the process of converting **JSON-formatted data** into **Python objects**.

### `json.loads()` – Convert JSON String to Python Object

The **`json.loads()`** function is used to convert a **JSON-formatted string** into a **Python object**.

````python
json_data = '{"name": "Alice", "age": 25, "city": "New York"}'

python_obj = json.loads(json_data)
print(python_obj)  
print(type(python_obj))  
````

## `json.load()` – Read JSON Data from a File

The **`json.load()`** function is used to read **JSON data** from a file and convert it into a **Python object**.

````python
with open("data.json","r")as file:
    py_data=json.load(file)
print(py_data)    
````

## Formatting JSON Output

You can format **JSON** for better readability by using **indentation**.

````python
formatted_json = json.dumps(data, indent=4)
print(formatted_json)
````
