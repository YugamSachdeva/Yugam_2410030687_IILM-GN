## Question 8: Find the Second Largest Number

Write a Python program to find the second largest unique number in a list.

Do not use `sort()` or `max()`.

**Example:**

Input:
[10, 5, 20, 8, 20, 15]

Expected Output:
15

```python
lst=[10, 5, 20, 8, 20, 15]
lst=list(set(lst))
for i in range(0,len(lst)):
    for j in range(i+1,len(lst)):
        if lst[i]>lst[j]:
            temp=lst[i]
            lst[i]=lst[j]
            lst[j]=temp

print("Second largest no:",lst[-2])

```

## Question 11: Check for Anagram

Write a Python program to check whether two strings are anagrams or not.

Two strings are anagrams if they contain the same characters with the same frequency, but the order can be different.

Ignore spaces and uppercase/lowercase differences.

**Example:**

Input:
"listen"
"silent"

Expected Output:
"Anagram"

Input:
"hello"
"world"

Expected Output:
"Not Anagram"

```python
s1="Silent"
s2="Listen"
ds1={}
ds2={}
for ch in s1.lower().replace(" ", ""):
    if ch in ds1:
        ds1[ch]+=1
    else:
        ds1[ch]=1
for ch in s2.lower().replace(" ", ""):
    if ch in ds2:
        ds2[ch]+=1
    else:
        ds2[ch]=1        
if ds1==ds2:
    print("Anagram")
else:
    print("Not Anagram")
```

## Question 17: Merge Two Sorted Lists

Write a Python program to merge two already sorted lists into one sorted list.

Do not use `sort()` or `sorted()`.

**Example:**

Input:
list1 = [1, 3, 5, 7]
list2 = [2, 4, 6, 8]

Expected Output:
[1, 2, 3, 4, 5, 6, 7, 8]

```python
list1 = [1, 3, 5, 7] 
list2 = [2, 4, 6, 8]
merge=[]        
i,j=0,0
while i != len(list1) and j != len(list2):
    if list1[i] < list2[j]:
        merge.append(list1[i])
        i += 1
    else:
        merge.append(list2[j])
        j += 1
while i < len(list1):
    merge.append(list1[i])
    i+=1
while j< len(list2):
    merge.append(list2[j])
    j+=1
print(merge)        
```

## Question 29: Safe Division

Write a Python program that takes two numbers from the user and divides the first number by the second.

Handle the case where the user enters 0 as the denominator using `try-except`.

If division is successful, print the result.

If the denominator is 0, print:
"Cannot divide by zero"

Example:

Input:
10
2

Expected Output:
5.0

Input:
10
0

Expected Output:
Cannot divide by zero

Conditions:
- Must use try-except.
- Handle ZeroDivisionError specifically.

```python
#Exception Handling
try: 
    a=int(input("Enter 1st number:"))
    b=int(input("Enter 2nd number:"))
    output=a/b
    print(output)
except ZeroDivisionError:
    print("Can't divide by zero")
except ValueError:
    print("You have entered wrong datatype.")
else:
    print("This code is correct.")
finally:
    print("code successfully executed.")
```

## Question 30: Student Data Analysis

Given the following list of dictionaries:

students = [
    {"name": "Yugam", "marks": 85},
    {"name": "Rahul", "marks": 92},
    {"name": "Aman", "marks": 78},
    {"name": "Rohit", "marks": 95},
    {"name": "Karan", "marks": 88}
]

Write a Python program to:

1. Find the average marks.
2. Find the student with the highest marks.
3. Create a list containing only students who scored 85 or more.

Expected Output:

Average: 87.6
Top Student: Rohit - 95
Students with 85+: 
["Yugam", "Rahul", "Rohit", "Karan"]

Conditions:
- Do not use max().
- Use loops/dictionaries/lists.
- For the last part, you may use list comprehension.

```python
from functools import reduce
students = [ {"name": "Yugam", "marks": 85}, {"name": "Rahul", "marks": 92}, {"name": "Aman", "marks": 78}, {"name": "Rohit", "marks": 95}, {"name": "Karan", "marks": 88} ]
refined_dict={}
for dict in students:
    refined_dict[dict["name"]]=dict["marks"]  
avg=(reduce(lambda a,b:a+b,refined_dict.values()))/len(refined_dict)  
print(f"Average of total marks:{avg}")
highest_marks=0   
for i in refined_dict.values():
    if i>highest_marks:
        highest_marks=i
for n,m in refined_dict.items():
    if m==highest_marks:
        print(f"{n} scored highest marks={m}.")
lst_85=[]
for n,m in refined_dict.items():
    if m>=85:
        lst_85.append(n)
print(f"Student who scored more than 85 are {lst_85}")        
```
