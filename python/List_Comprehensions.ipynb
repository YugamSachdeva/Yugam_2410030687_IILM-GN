# List Comprehensions

**List comprehension** is a concise and efficient way to create **lists** in Python. It allows you to generate lists in a **single line of code**, making your code more **readable**, **concise**, and **Pythonic**.

## 1. Basic Syntax

```python
[expression for item in iterable]
```

- **expression** → The operation to perform on each item.
- **item** → The variable representing each element in the iterable.
- **iterable** → The data structure being iterated over (such as a list, tuple, or `range()`).

### Example: Creating a List of Squares

````python
list=[x**2 for x in range(5)]
print(list)
````

## 2. Using `if` Condition in List Comprehension

You can use an **`if` condition** in a list comprehension to include only those elements that satisfy a specific condition.

### Example: Filtering Even Numbers

````python
list=[x for x in range(11) if x%2==0]
print(list)
````

## 3. Using `if-else` Condition in List Comprehension

You can use an **`if-else` condition** in a list comprehension to perform different operations based on a condition.

### Example: Replacing Even Numbers with `"Even"` and Odd Numbers with `"Odd"`

````python
list=["EVEN" if x%2==0 else "ODD" for x in range(6)]
print(list)
````

## 4. Nested Loops in List Comprehension

You can use **nested loops** in a list comprehension to iterate over multiple iterables.

### Example: Creating Pairs from Two Lists

````python
pairs = [(x, y) for x in range(2) for y in range(3)]
print(pairs)  
````

## 5. List Comprehension with Functions

You can use **functions** inside a list comprehension to perform operations on each element.

### Example: Converting a List of Strings to Uppercase

````python
word=["hey","what's up?","insane"]
upper_word=[words.upper() for words in word]
print(upper_word)
````

## 6. List Comprehension with Nested List Comprehension

You can use a **nested list comprehension** to work with **nested lists** or perform operations on multi-dimensional data.

### Example: Flattening a 2D List

````python
matrix = [[1, 2], [3, 4], [5, 6]]
flattened=[num for row in matrix for num in row]
print(flattened)
````

## 7. List Comprehension with Set and Dictionary Comprehensions

### Set Comprehension

A **set comprehension** is used to create a **set** in a single line of code, similar to a list comprehension.

````python
unique_numbers = {x for x in [1, 2, 2, 3, 4, 4]}
print(unique_numbers)  # Output: {1, 2, 3, 4}
````

### Dictionary Comprehension

A **dictionary comprehension** is used to create a **dictionary** in a single line of code using key-value pairs.

````python
squared_dict = {x: x**2 for x in range(5)}
print(squared_dict)  # Output: {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}
````

## 8. When to Use List Comprehensions?

Use list comprehensions when:

- You need to create a list in a single line.
- The logic is simple and easy to read.
- You want better performance (generally faster than traditional loops).

### Avoid When

- The logic is too complex (use a standard loop instead for better readability).
