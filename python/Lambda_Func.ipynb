# Lambda Functions

A **lambda function** in Python is an **anonymous, single-expression function** defined using the **`lambda`** keyword. It is commonly used for **short, throwaway functions** where a full function definition is unnecessary.

---

## 1. Syntax of Lambda Functions

```python
lambda arguments: expression
```

- **`lambda`** → Keyword used to define a lambda function.
- **`arguments`** → Input parameters (comma-separated).
- **`expression`** → The operation performed (must be a **single expression**, not multiple statements).

### Example: Simple Lambda Function

````python
cube=lambda x:x*x*x
print(cube(3))
````

## 2. Using Lambda Functions with `map()`, `filter()`, and `reduce()`

### 2.1 Using `map()` with Lambda

The **`map()`** function applies a given function to **each element** of an iterable and returns a **map object**. It is commonly used with **lambda functions** for concise and efficient code.

````python
numbers=[1,2,3,4,5]
square=list(map(lambda x:x**2,numbers))
print(square)
````

## 2.2 Using `filter()` with Lambda

The **`filter()`** function filters elements from an iterable based on a **condition** and returns only the elements that satisfy the condition.

````python
number=[1,2,3,4,5,6]
even=list(filter(lambda x:x%2==0,number))
print(even)
````

## 2.3 Using `reduce()` with Lambda

The **`reduce()`** function reduces an iterable to a **single value** by repeatedly applying a function to its elements. It requires the **`functools.reduce()`** function.

````python
from functools import reduce

numbers = [1, 2, 3, 4]
product = reduce(lambda x, y: x * y, numbers)
print(product)  
````

## 3. Lambda with Multiple Arguments

A **lambda function** can accept **multiple arguments**, just like a normal function.

### Example: Adding Two Numbers

````python
add=lambda x,y:x+y
print(add(4,5))
````

### Example: Finding the Maximum of Two Numbers

````python
max_no=lambda x,y:x if x>y else y
print(max_no(5,6))
````

## 4. Lambda in Sorting Functions

Lambda functions are commonly used with **sorting functions** to specify a custom sorting key.

### Example: Sorting a List of Tuples

````python
students = [("Alice", 85), ("Bob", 78), ("Charlie", 92)]
students.sort(key=lambda student: student[1]) 
print(students)  
````
